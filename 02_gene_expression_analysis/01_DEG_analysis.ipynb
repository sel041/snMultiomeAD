## Function for DEG analysis

```r
all.paths <- list.files('data/Celltype_Seurat_Objects/')
splitted.objects <- vector(mode = 'list', length = length(all.paths))
for (path in all.paths) {
    celltype <- strsplit(path, split = '[.]')[[1]][1]
    print(celltype)
    flush.console()
    splitted.objects[[celltype]] <- readRDS(file.path('data//Celltype_Seurat_Objects', path))
}
```

### The function

```r
edgeR.analysis <- function(data.objects, celltype.name) {
    curr.object <- data.objects[[celltype.name]]
    curr.counts <- LayerData(curr.object, layer = c('counts'))
    curr.meta <- curr.object@meta.data
    
    curr.sce <- SingleCellExperiment(assays = list(counts = curr.counts), colData = curr.meta)
    
    curr.sce$age <- as.character(curr.sce$age)
    curr.sce$age <- factor(curr.sce$age, levels = c('3M', '9M', '18M'))
    curr.sce$genotype <- as.character(curr.sce$genotype)
    curr.sce$genotype[curr.sce$genotype == '5XFAD'] <- 'AD'
    curr.sce$genotype <- factor(curr.sce$genotype, levels = c('WT', 'AD'))
    curr.sce$intervention <- as.character(curr.sce$intervention)
    curr.sce$intervention[curr.sce$intervention == 'na'] <- 'Sed'
    curr.sce$intervention <- factor(curr.sce$intervention, levels = c('Sed', 'Ex'))
    curr.sce$sex <- as.character(curr.sce$sex)
    curr.sce$sex <- factor(curr.sce$sex, levels = c('M', 'F'))
    
    groups <- curr.sce$mouse_id
    curr.aggr <- aggregateAcrossCells(curr.sce, groups, store.number = 'group.size')
    
    Genotype <- curr.aggr$genotype
    Age <- curr.aggr$age
    Intervention <- curr.aggr$intervention
    Sex <- curr.aggr$sex
    
    ## Three factor model (no intervention)
    Group <- paste(Genotype, Age, Sex, sep = '.')
    Group <- factor(
        Group, 
        levels = c(
            'WT.3M.M', 'WT.3M.F',
            'AD.3M.M', 'AD.3M.F',
            'WT.9M.M', 'WT.9M.F',
            'AD.9M.M', 'AD.9M.F',
            'WT.18M.M', 'WT.18M.F',
            'AD.18M.M', 'AD.18M.F'
        )
    )
    
    curr.dge <- DGEList(counts = counts(curr.aggr), group = Group, remove.zeros = TRUE)
    keep <- filterByExpr(curr.dge, min.count = 5, min.prop = 0.5)
    curr.dge <- curr.dge[keep, , keep.lib.sizes=FALSE]
    curr.dge <- calcNormFactors(curr.dge, method = 'TMM')
    
    options(repr.plot.width = 8, repr.plot.height = 8)
    edgeR::plotMDS.DGEList(curr.dge, labels = Group, main = celltype.name)
    
    curr.design <- model.matrix(~ 0 + Group)
    colnames(curr.design) <- levels(Group)
    
    curr.dge <- estimateDisp(curr.dge, curr.design, robust = TRUE)
    
    edgeR::plotBCV(curr.dge, main = paste(celltype.name, "No Intervention", sep = " "))
    
    fit <- glmQLFit(curr.dge, curr.design)
    saveRDS(fit, file.path('data/DGE_fit_no_intervention_models/', paste(celltype.name, 'model', sep = '.')))
    
    
    ## Three factor model (no sex)
    Group <- paste(Genotype, Age, Intervention, sep = '.')
    Group <- factor(
        Group, 
        levels = c(
            'WT.3M.Sed', 'AD.3M.Sed',
            'WT.9M.Sed', 'WT.9M.Ex',
            'AD.9M.Sed', 'AD.9M.Ex',
            'WT.18M.Sed', 'WT.18M.Ex',
            'AD.18M.Sed', 'AD.18M.Ex'
        )
    )
    
    curr.dge <- DGEList(counts = counts(curr.aggr), group = Group, remove.zeros = TRUE)
    keep <- filterByExpr(curr.dge, min.count = 5, min.prop = 0.5)
    curr.dge <- curr.dge[keep, , keep.lib.sizes=FALSE]
    curr.dge <- calcNormFactors(curr.dge, method = 'TMM')
    # edgeR::plotMDS.DGEList(curr.dge, labels = Group)
    
    curr.design <- model.matrix(~ 0 + Group)
    colnames(curr.design) <- levels(Group)
    
    curr.dge <- estimateDisp(curr.dge, curr.design, robust = TRUE)
    edgeR::plotBCV(curr.dge, main = paste(celltype.name, "No Sex", sep = " "))
    
    fit <- glmQLFit(curr.dge, curr.design)
    saveRDS(fit, file.path('data/DGE_fit_no_sex_models/', paste(celltype.name, 'model', sep = '.')))
    
    
    ## Two factor model (age and genotype)
    Group <- paste(Genotype, Age, sep = '.')
    Group <- factor(
        Group, 
        levels = c('WT.3M', 'AD.3M', 'WT.9M', 'AD.9M', 'WT.18M', 'AD.18M')
    )
    
    curr.dge <- DGEList(counts = counts(curr.aggr), group = Group, remove.zeros = TRUE)
    keep <- filterByExpr(curr.dge, min.count = 5, min.prop = 0.5)
    curr.dge <- curr.dge[keep, , keep.lib.sizes=FALSE]
    curr.dge <- calcNormFactors(curr.dge, method = 'TMM')
    # edgeR::plotMDS.DGEList(curr.dge, labels = Group)
    
    curr.design <- model.matrix(~ 0 + Group)
    colnames(curr.design) <- levels(Group)
    
    curr.dge <- estimateDisp(curr.dge, curr.design, robust = TRUE)
    edgeR::plotBCV(curr.dge, main = paste(celltype.name, "Genotype X Age", sep = " "))
    
    fit <- glmQLFit(curr.dge, curr.design)
    saveRDS(fit, file.path('data/DGE_fit_genotype_age_models/', paste(celltype.name, 'model', sep = '.')))
}
```

### Iterate through all cell types

```r
all.paths <- list.files('data/Celltype_Seurat_Objects/')
for (path in all.paths) {
    celltype <- strsplit(path, split = '[.]')[[1]][1]
    print(celltype)
    flush.console()
    if (celltype %in% c('SMC-Peri')) {
        next
    }
    edgeR.analysis(splitted.objects, celltype)
}
```
